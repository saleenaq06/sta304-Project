```python
import pandas as pd

df = pd.read_csv('studentresponse.csv')

# prints the first 5
df.head(5)
```

```python
df.columns.tolist()
```

### Column 1, 2, 3, 4

```python
# Removes first column: 'Timestamp'
df = df.drop(columns='Timestamp')
df.head(2)
```

```python
# 'Do you consent to the data you provide to be used for statistical and research purposes?'
df.rename(columns={'Do you consent to the data you provide to be used for statistical and research purposes?': 'consent'}, inplace=True)
df.head(2)
```

```python
# Renames students’ unique identifiers EXACTLY as “studentID”
df.rename(columns={'What is your assigned course specific ID?': 'studentID'}, inplace=True)
df.head(2)
```

```python
# Maps the column 'What is your current year of study?' to categorical integers
year_mapping = {
    '1st year': 1,
    '2nd year': 2,
    '3rd year': 3,
    '4th year': 4,
    '5th year or higher': 5,
}

df['What is your current year of study?'] = df['What is your current year of study?'].map(year_mapping)

df.head(2)
```

### Column 5, 6 Saleena

```python
# 'What is your current student status?'

# Changes name of column 'What is your current student status?' to 'Citizenship Status'
df.rename(columns={'What is your current student status?': 'Citizenship Status'}, inplace=True)
df.head(2)
```

```python
# Get rid of "student" after domestic or international in the column 'Citizenship Status'

df['Citizenship Status'] = df['Citizenship Status'].str.replace('Domestic student', '0')
df['Citizenship Status'] = df['Citizenship Status'].str.replace('International student', '1')

df.head(4)
```

```python
# 'How many connections do you have on LinkedIn?'

# Renaming the column 'How many connections do you have on LinkedIn?' to 'LinkedIn Connections'

df.rename(columns={'How many connections do you have on LinkedIn?': 'LinkedIn Connections'}, inplace=True)

linkedin_connections_map = {
    'I do not use LinkedIn or Prefer not to say': 0,
    '0 - 100': 1,
    '100 - 200': 2,
    '200 - 300': 3,
    '300 - 400': 4,
    '400 - 500': 5,
    '500+': 6
}
df['LinkedIn Connections'] = df['LinkedIn Connections'].map(linkedin_connections_map)

df.head(2)
```

### Column 7, 12, 13 Tanya

```python
# 'What is your current cumulative GPA?'

# Rename the column to 'CGPA'
df.rename(columns={'What is your current cumulative GPA?': 'CGPA'}, inplace=True)
df.head(2)
```

```python
cgpa_map = {
    'Below 1.70': 0,
    '1.70 - 2.19': 1,
    '2.20 - 2.69': 2,
    '2.70 - 3.19': 3,
    '3.20 - 3.69': 4,
    '3.7+': 5}

df['CGPA'] = df['CGPA'].map(cgpa_map)
df.head(2)
```

```python
# 'How many internships (paid or unpaid, including co-op placements, ROP, and UTMCIP) have you completed?'

# Renaming the column 'How many internships (paid or unpaid, including co-op placements, ROP, and UTMCIP) have you completed?' to 'Completed Internships'

df.rename(columns={'How many internships (paid or unpaid, including co-op placements, ROP, and UTMCIP) have you completed?': 'Completed Internships'}, inplace=True)
#Convert internship responses to numeric values

df['Completed Internships'] = df['Completed Internships'].replace('12 months of PEY', 1)
df['Completed Internships'] = pd.to_numeric(df['Completed Internships'], errors='coerce').astype('Int64')
df.head(2)
```

```python
# 'How well do you feel your degree has prepared you for the workforce?'

# Rename the column to 'Workforce Preparedness'
df.rename(columns={'How well do you feel your degree has prepared you for the workforce?': 'Workforce Preparedness'}, inplace=True)

# Map workforce preparedness responses to numerical values
workforce_preparedness_map = {
    'Strongly prepared': 1,
    'Moderately prepared': 2,
    'Slightly prepared': 3,
    'Neutral / Undecided': 4,
    'Slightly unprepared': 5,
    'Moderately unprepared': 6,
    'Strongly unprepared': 7
}

df['Workforce Preparedness'] = df['Workforce Preparedness'].map(workforce_preparedness_map)
df.head(2)
```

### Column 8, 9 Krish

```python
# 'Rate your agreement with each statement [I have the necessary skills to compete for an entry level job position.]'

# Rename the column to 'Necessary Skills'
df.rename(columns={'Rate your agreement with each statement [I have the necessary skills to compete for an entry level job position.]': 'Necessary Skills'}, inplace=True)

# Code agreement levels (same coding as Alisyee's columns: 1 = Strongly Agree, 7 = Strongly Disagree)
agreement_map = {
    'Strongly Agree': 1,
    'Agree': 2,
    'Slightly Agree': 3,
    'Neutral': 4,
    'Slightly Disagree': 5,
    'Disagree': 6,
    'Strongly Disagree': 7,
}

df['Necessary Skills'] = df['Necessary Skills'].str.strip().map(agreement_map)
df.head(2)
```

```python
# 'Rate your agreement with each statement [I believe there are abundant entry-level opportunities in my field of study right now.]'

# Rename the column to 'Abundant Opportunities'
df.rename(columns={'Rate your agreement with each statement [I believe there are abundant entry-level opportunities in my field of study right now.]': 'Abundant Opportunities'}, inplace=True)

df['Abundant Opportunities'] = df['Abundant Opportunities'].str.strip().map(agreement_map)

# Check: no missing values and every value is between 1 and 7
print(df[['Necessary Skills', 'Abundant Opportunities']].isna().sum())
assert df[['Necessary Skills', 'Abundant Opportunities']].isin(range(1, 8)).all().all()
df.head(2)
```
